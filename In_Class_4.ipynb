# Exercise 3 (8 points)

How many products cost more than $10.00?

```python
chipotle["item_price"] = chipotle["item_price"].str.replace("$", "").astype(float)
chipotle.head()
```

```python
chipotle.info()
```

```python
chipotle[(chipotle["quantity"] == 1) & (chipotle["item_price"] > 10)]
```

```python
chipotle[(chipotle["quantity"] == 1) & (chipotle["item_price"] > 10)]["item_name"].unique()
```

```python
chipotle[(chipotle["quantity"] == 1) & (chipotle["item_price"] > 10)]["item_name"].nunique()
```
