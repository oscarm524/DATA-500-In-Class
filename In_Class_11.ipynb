Chapter 9 describes a **hierarchy of documentation** — names, type hints, comments, docstrings, API documentation, and READMEs — and notes that this hierarchy is ordered by *cost* and by *distance from the code*.

Answer the following:

1. Explain what it means for documentation to be "close to the code" versus "far from the code," and why the chapter recommends pushing information as far up the hierarchy (toward names and type hints) as it will go.
2. Explain the difference in purpose between a **comment** and a **docstring**.
3. The chapter states that "duplicated documentation is worse than missing documentation." Explain why this is true, and give a concrete example of how this problem could occur in a data science project (e.g., a parameter description that lives in both a docstring and a README).

1. Documetation that is "close to the code" lives directly in the function or class signature, so the reader sees it in the glance as the logic itself. Documentation that is "far from the code" (a README, a wiki page) lives in separate file that is not checked by any tool, so it can silently drift out of sync as the code changes. It is recommended to push information as far up the hierarchy as possible becuae the layers closer to the code are cheaper to maintain and cannot "silently disagree" with the code.

2. A comment is an inline, lightweight note meant to explain why a specific piece of code does something non-obvious. A docstring is the formal, structured documentation of a function, class, or module. It systematically describes the function's behavior, parameters, return value, exceptions, and edge cases. 

3. Duplicated documentation is worse than missing documentation because when the same fact is written in two or more places, there is no mechanism keeping those copies synchronized. Sooner or later the code changes and only one copy is updated, and now the reader has no way to know which version is correct. Example: a parameter's valid range for an argument is documented both in the function's docstring ("must be non-negative") and in the project's README. If the function is later changed to allowed negative thresholds, and the docstring is updated, the README now contradicts the actual behavior, and a new user following the README may pass a value that disagrees with what the code now expects.

In this exercise, we will practice documenting code the way Chapter 9 describes: with a descriptive name, type hints, and a complete Google-style docstring (summary, `Args`, `Returns`, `Raises`, and `Example`).

Consider the following undocumented, poorly named function:

```python
def f(df, n):
    df = df[df['price'] >= n]
    df['price'] = df['price'].astype(float)
    return df
```

Your task:

- **Rename the function** to something descriptive that follows the `verb_noun` convention (e.g., a function that filters rows should start with a verb like `filter_`).
- **Add type hints** to the parameters and the return value (the input and output are `pandas.DataFrame`; the threshold is a `float`).
- **Add a `ValueError`** that is raised if the threshold `n` is negative.
- **Write a complete Google-style docstring** including a one-line summary, `Args`, `Returns`, `Raises`, and a runnable `Example`.

Then apply your documented function to the following data, using a minimum price threshold of `20`.

```python
import pandas as pd

data = {'item': ['A', 'B', 'C', 'D'],
        'price': ['10', '25', '30', '15']}
df = pd.DataFrame(data)
```

The expected outcome is shown below.

| item | price |
| --- | --- |
| B | 25.0 |
| C | 30.0 |

````python
import pandas as pd 

def filter_by_min_price(df: pd.DataFrame, min_price: float) -> pd.DataFrame:
    """Filter rows to prices greater than or equal to the specified minimum price.

    Args:
        df (pd.DataFrame): Input data containing at least a 'price' column.
        min_price (float): Minimum price to filter the rwos by.

    Returns:
        pd.DataFrame: Filtered DataFrame containing only rows with 'price' greater than or equal to min_price.

    Raises:
        ValueError: If min_price is negative.

    Example:
        >>> df = pd.DataFrame({'price': [10, 20, 30]})
        >>> filter_by_min_price(df, 20)
           price
        1     20
        2     30
    """
    if min_price < 0:
        raise ValueError("min_price cannot be negative")
    
    df = df.copy()
    df['price'] = df['price'].astype(float)
    return df[df['price'] >= min_price]


data = {'item': ['A', 'B', 'C', 'D'],
        'price': ['10', '25', '30', '15']}
df = pd.DataFrame(data)    

filtered_df = filter_by_min_price(df, 20)
print(filtered_df)
````
